Q.2
a customer want to purchase multiple products . you are supposed to ask how many products they want to buy and then collect the product name , quantity and price

```python
# Q2 solution
def user_purchase():
    name = input("Please enter your name: ")
    product_count = int(input("How many products do you want to buy? "))

    cart = []
    total_bill = 0

    for i in range(product_count):
        print(f"\nEntering details for product {i+1}:")
        p_name = input("Enter product name: ")
        quantity = float(input("Enter quantity: "))
        price = float(input("Enter price: "))

        item_total = quantity * price
        total_bill += item_total

        cart.append({"name": p_name, "qty": quantity, "price": price, "total": item_total})

    print(f"\n--- Invoice for {name} ---")
    for item in cart:
        print(f"{item['name']}: {item['qty']} x {item['price']} = {item['total']}")
    print(f"Total Bill Amount: {total_bill}")

user_purchase()
```
