```python

text = " "
undo_stack = []

from typing_extensions import Text
def type_text(new_text):
  global text

  #  save current text befre making a change
  undo_stack.append(text)

  # add new text
  text = text+new_text
  print("Text: ", text)


def undo():
  global text


  # SAVE CURRENT TEXT BEFORE making a change
  undo_stack.append(text)


  # add new text
  text = text + new_text

  print("Text:", text)


def undo ():
  global text

  if undo_stack:
    text = undo_stack.pop()
    print("Undo:", text)

  else:
    print("Nothing to undo")

# try the editor
type_text("Hello ")
type_text("World")
type_text("!")

print("\nPerforming Undo opetations:")
undo()
undo()
undo()
undo()
```
