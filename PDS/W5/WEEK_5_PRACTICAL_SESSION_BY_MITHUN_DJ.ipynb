```python
# DATA STRUCTURES - STACK AND QUEUES


# ACTIVITY 1 : BASIC STACK
# LIFO = Last In, First out

stack = []

# add items

stack.append("A")
stack.append("B")
stack.append("C")
print("Stack:", stack)
```

```python
# SEE ONT HTE TOP
print("Top ites:", stack[-1])
```

```python
# remove the top item
removed = stack.pop()
print("Removed:", removed)

print("Stack after pop", stack)
```
