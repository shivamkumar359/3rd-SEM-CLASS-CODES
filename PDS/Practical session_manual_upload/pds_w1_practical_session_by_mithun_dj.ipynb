```python
# LITERALS
# ----------------------

integer_literal = 100
float_literal = 99.95
string_literal = "Python"
boolean_literal = True
```

```python
print("Integer literal :", integer_literal)
print("Float literal ;", float_literal)
print("Boolean literal :", boolean_literal)
print("String literal :", string_literal)
```

# 3. DATA TYPES


```python
print("Data types")
print(type(integer_literal))
print(type(float_literal))
print(type(string_literal))
print(type(boolean_literal))
```

```python
# Check if a number is even or odd

number = input("Enter a number: ")

print("Before conversion:", type(number))

number = int(number)

print("After conversion:", type(number))

if number % 2 == 0:
    result = "even"
else:
    result = "odd"

print("The number is {}.".format(result))
```
