# 4. Looping

```python
# 1 While loop
i = 0
while i < 10:
  i += 1
  print(i)


  if i == 10:
    break
```

```python
# 2. While loop
i = 1
while i < 6:
  print(i)

if i == 3:
  break
i += 1
```

```python
# 3. For loop
for i in range(10):
  print(i)      # 0,1,2,3,....,9

for i in range(3,10):
  print(i)      # 3,4,5,6.....,9

for i in range(0,10,2):
  print(i)      # 0,2,4,6,8

else:
  print("Loop completed ")
```
