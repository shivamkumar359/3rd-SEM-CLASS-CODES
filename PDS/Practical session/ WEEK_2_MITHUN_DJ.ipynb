```python
# 4. BUbble sort method


numbers = [53,8,78,1,245,554,12,34,5,2]
print("Before:",numbers)
```

```python
n = len(numbers)
# Outer loop = number of passes
for i in range(n):

  # inner loop = compare neighbouting elements
  for j in range(0, n-i-1):

    # if left number is bigger
    if numbers[j] > numbers[j+1]:

      # swap
      numbers[j], numbers[j+1] = numbers[j+1], numbers[j]

print("After", numbers)
```

```python
# ============================================================
# PART 5: BUBBLE SORT - SHOW EVERY PASS
# ============================================================
print("\n\nBUBBLE SORT - STEP BY STEP")
print("=" * 40)
numbers = [5, 3, 8, 1, 2]
print("Starting list:", numbers)
n = len(numbers)
for i in range(n):
    print("\nPass", i + 1)
    for j in range(0, n - i - 1):
        print("Compare:", numbers[j], "and", numbers[j + 1])
        if numbers[j] > numbers[j + 1]:
            print("SWAP")
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
        else:
            print("NO SWAP")
        print("List:", numbers)
print("\nFinal sorted list:", numbers)
```
