```python
import pandas as pd

teams_data = {
    'team_id': [1, 2, 3],
    'team_name': ['MI', 'CSK', 'RCB'],
    'city': ['Mumbai', 'Chennai', 'Bangalore'],
    'coach_name': ['Mark', 'Stephen', 'Andy']
}
teams = pd.DataFrame(teams_data)

players_data = {
    'player_id': [101, 102, 103, 104, 105, 106, 107, 108],
    'player_name': ['Rohit Sharma', 'Jasprit Bumrah', 'Hardik Pandya', 'MS Dhoni', 'Ravindra Jadeja', 'Virat Kohli', 'Mohammed Siraj', 'Glenn Maxwell'],
    'age': [36, 30, 30, 42, 35, 35, 29, 35],
    'role': ['batsman', 'bowler', 'all-rounder', 'wicket-keeper', 'all-rounder', 'batsman', 'bowler', 'all-rounder'],
    'team_id': [1, 1, 1, 2, 2, 3, 3, 3]
}


players = pd.DataFrame(players_data)

print("Team Table:")
display(teams)
print("\nPlayers Table:")
display(players)
```

```python
young_players = players[players['age'] < 31]
display(young_players)
```

```python
young_bowlers = players[(players['role'] =='bowler') & (players['age'] < 31)]
display(young_bowlers)
```
