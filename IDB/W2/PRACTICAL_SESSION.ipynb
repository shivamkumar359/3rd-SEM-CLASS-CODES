```python
import pandas as pd


teams_data = {
    'team_id': [1, 2, 3],
    'team_name': ['Mumbai Indians', 'Chennai Super Kings', 'Royal Challengers Bangalore'],
    'city': ['Mumbai', 'Chennai', 'Bangalore'],
    'coach_name': ['Mark Boucher', 'Stephen Fleming', 'Andy Flower']
}
teams = pd.DataFrame(teams_data)

players_data = {
    'player_id': [101, 102, 103, 104, 105, 106, 107, 108],
    'player_name': ['Rohit Sharma', 'Jasprit Bumrah', 'Hardik Pandya', 'MS Dhoni', 'Ravindra Jadeja', 'Virat Kohli', 'Mohammed Siraj', 'Glenn Maxwell'],
    'age': [36, 30, 30, 42, 35, 35, 29, 35],
    'role': ['batsman', 'bowler', 'all-rounder', 'wicket-keeper', 'all-rounder', 'batsman', 'bowler', 'all-rounder'],
    'team_id': [1, 1, 1, 2, 2, 3, 3, 3]
}
players = pd.DataFrame(players_data)

print("Teams Table:")
display(teams)
print("\nPlayers Table:")
display(players)
```

```python
young_players = players[players['age'] < 32]
display(young_players)
```

```python
young_bowlers = players[(players['role'] =='bowler') & (players['age'] < 31)]
display(young_bowlers)
```

```python
all_rounders_info = players[players['role'] == 'all-rounder'][['player_name', 'age']]
display(all_rounders_info)
```

## Part C — Set Operations
Union, Intersection, and Set Difference using the MI 2023 and 2024 squads.

```python
mi_2023 = pd.DataFrame({'player_name': ['Rohit Sharma', 'Jasprit Bumrah', 'Jofra Archer'], 'role': ['batsman', 'bowler', 'bowler']})
mi_2024 = pd.DataFrame({'player_name': ['Rohit Sharma', 'Jasprit Bumrah', 'Hardik Pandya'], 'role': ['batsman', 'bowler', 'all-rounder']})

union_df = pd.concat([mi_2023, mi_2024]).drop_duplicates()
intersection_df = pd.merge(mi_2023, mi_2024, how='inner')
merged = mi_2023.merge(mi_2024, on=['player_name', 'role'], how='left', indicator=True)
set_diff_df = merged[merged['_merge'] == 'left_only'].drop('_merge', axis=1)

print("UNION (All Players):"); display(union_df)
print("\nINTERSECTION (Loyalists):"); display(intersection_df)
print("\nSET DIFFERENCE (Released Players):"); display(set_diff_df)
```

## Part D + E — Join and Aggregation
Cartesian products, Equality conditions, and Aggregations (Count, Avg, Max).

```python
cartesian_product = players.merge(teams, how='cross')
print(f"Total rows: {len(cartesian_product)}")
display(cartesian_product.head(10))
```

```python
joined_df = pd.merge(players, teams, on='team_id')
print(f"Total rows: {len(joined_df)}")
display(joined_df)
```

```python
pipeline_result = pd.merge(players, teams, on='team_id') \
                    [players['role'] == 'bowler'] \
                    [['player_name', 'city']] \
                    .rename(columns={'player_name': 'Bowler Name', 'city': 'Franchise City'})
display(pipeline_result)
```

```python
players_per_team = joined_df.groupby('team_name')['player_id'].count().reset_index(name='Player Count')
avg_age = joined_df.groupby('team_name')['age'].mean().reset_index(name='Average Age')
oldest = players[players['age'] == players['age'].max()]

print("COUNT:"); display(players_per_team)
print("\nAVG:"); display(avg_age)
print("\nMAX:"); display(oldest)
```
