```python
marks = [78,98,56,29,66,92,88,65]
```

```python
# Remove a value
marks.remove(66)
print("After remove: ", marks)
```

```python
# remove last items
marks.pop()
print("After pop:", marks)
```

```python
# sort
marks.sort()
print("Sorted marks:", marks)
```

```python
# reverse sort
marks.sort(reverse=False)

print("Ascending sorted marks:", marks)
```

```python
# reverse sort
marks.sort(reverse=True)

print("Descending sorted marks:", marks)
```

```python
# length
print("Length of marks:", len(marks))
```

```python
# total and average
print("Total:", sum(marks))
```

```python
print("Average:", sum(marks)/len(marks))
```

```python
# tuple is ordered but immutable
student = ("Rahul", 21,"Data Science")
print("Tuple:", student)
```

```python
# Access
print("Name:",student[0])
print("Age:", student[1])
print("Course:", student[2])
```

```python
# Slicing
print("First two values:", student[:2])
```

```python
#  Tuple unpacking
name , age, course = student

print("Name:", name)
print("Age:", age)
print("Course:", course)
```

```python
# dictionary stores key:values paris
student_marks = {
    "Shiva":87,
    "Priya":91,
    "Arun":56,
    "Rahul":78
  }
```

```python
print("Dictionary:", student_marks)
```

```python
# Access using key
print("Priya's marks:", student_marks["Priya"])
```

```python
# safer method using get()
print("Rahul's marks:", student_marks.get("Rahul"))
```

```python
# safer method using get()
print("Kiran's marks:",
                  student_marks.get("Kiran", "Kiran not found"))
```

```python
# add new student
student_marks["Kiran"]= 88
print("After updating Rahul:", student_marks)
```

```python
# update
student_marks["Rahul"]= 82
print("After updating Rahul:", student_marks)
```

```python
print("Stdnt names:", student_marks.keys())
```

```python
# values
print("Marks:",student_marks.values())
```

```python
# key value pairs
print("Marks:",student_marks.items())
```

```python
# check whetere a key exist or not
if "Priya" in student_marks:
  print("Priya is present")
```

```python
# Delete
student_marks.pop("Arun", None) # Use a default value to avoid KeyError if 'Arun' is not present
print("After deleting Arun:", student_marks)
```
